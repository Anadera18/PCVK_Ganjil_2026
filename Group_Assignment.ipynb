```python
img = cv.imread("20260830_160026.jpg.jpeg")

left_eye_center = (115, 680)
right_eye_center = (170, 680)
mouth_top_left = (120, 725)
mouth_bottom_right = (160, 735)

cv.circle(img, left_eye_center, radius=14, color=(0, 0, 255), thickness=3)
cv.circle(img, right_eye_center, radius=14, color=(0, 255, 0), thickness=3)
cv.rectangle(img, mouth_top_left, mouth_bottom_right, color=(255, 0, 0), thickness=3)

img_RGB = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))
plt.imshow(img_RGB)
plt.axis("on")
plt.show()
```
