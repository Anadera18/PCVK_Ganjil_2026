```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random

img = cv2.imread('20260830_160026.jpg (2).jpeg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- SOAL 1: Menggunakan figsize ---
plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title(f"1. Gambar Asli (Resolusi: {img_rgb.shape[1]}x{img_rgb.shape[0]})")
plt.show()
```

```python
# --- SOAL 2: Channel Red-Blue dan Green-Blue ---
# Red-Blue (Green dinolkan)
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0

# Green-Blue (Red dinolkan)
img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(img_rb)
ax[0].set_title('2a. Red-Blue Channel')
ax[1].imshow(img_gb)
ax[1].set_title('2b. Green-Blue Channel')
plt.show()
```

```python
crop_img = img_rgb[20:115, 25:120]
plt.figure(figsize=(4, 4))
plt.imshow(crop_img)
plt.title('3. Crop (Baris 20-115, Kolom 25-120)')
plt.show()
```

```python
# --- SOAL 4: Baris 5-30, semua kolom, Channel Red saja ---
# Agar citra tetap terlihat merah (bukan grayscale), kita buat matriks kosong dan hanya mengisi channel Red
img_red_only = np.zeros_like(img_rgb[5:30, :])
img_red_only[:, :, 0] = img_rgb[5:30, :, 0]

plt.figure(figsize=(10, 3))
plt.imshow(img_red_only)
plt.title('4. Crop Baris 5-30, Channel Red Saja')
plt.show()
```

```python
img_boxes = img_rgb.copy()
h, w, _ = img_boxes.shape

for _ in range(5):
    # Buat koordinat dan warna secara acak
    x1 = random.randint(0, w - 100)
    y1 = random.randint(0, h - 100)
    x2 = x1 + random.randint(30, 150)
    y2 = y1 + random.randint(30, 150)
    color = (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))
    thickness = random.randint(3, 8)

    cv2.rectangle(img_boxes, (x1, y1), (x2, y2), color, thickness)

plt.figure(figsize=(10, 6))
plt.imshow(img_boxes)
plt.title('5. Lima Kotak Random')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

filename = 'lena.jpg'

img_bgr = cv2.imread(filename)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

img_blue = img_rgb.copy()
img_blue[:, :, 0] = 0  # Red = 0
img_blue[:, :, 1] = 0  # Green = 0

h, w, _ = img_blue.shape
box_size = 100


cv2.rectangle(img_blue, (0, h - box_size), (box_size, h), (255, 255, 255), 4)

plt.figure(figsize=(7, 7))
plt.imshow(img_blue)
plt.axis('on')
plt.show()
```

```python
cv2.putText(img_blue, filename, (10, h - box_size - 15),
            cv2.FONT_HERSHEY_DUPLEX, 0.8, (0, 255, 255), 2, cv2.LINE_AA)
```
