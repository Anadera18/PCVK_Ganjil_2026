```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# 1. Load image and convert to RGB
img_orig = cv.imread('foto_aktifitas.jpg.jpeg')
img_orig = cv.cvtColor(img_orig, cv.COLOR_BGR2RGB)

h, w, _ = img_orig.shape

# 2. Calculate baseline average intensity
mean_orig = np.mean(img_orig)

# 3. Downscale to 25% of original dimensions
img_small = cv.resize(img_orig, (w // 4, h // 4), interpolation=cv.INTER_LINEAR)

# 4. Upscale back to original dimensions
img_restored = cv.resize(img_small, (w, h), interpolation=cv.INTER_LINEAR)

# 5. Calculate new average intensity
mean_restored = np.mean(img_restored)

# Calculate difference map
diff = cv.absdiff(img_orig, img_restored)

# Display results
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img_orig)
axes[0].set_title(f"Original\nMean: {mean_orig:.2f}")
axes[0].axis('off')

axes[1].imshow(img_restored)
axes[1].set_title(f"Resized & Restored\nMean: {mean_restored:.2f}")
axes[1].axis('off')

axes[2].imshow(diff)
axes[2].set_title(f"Absolute Difference\nMax Diff: {np.max(diff)}")
axes[2].axis('off')

plt.tight_layout()
plt.show()

# Print experimental metrics
print(f"Original Shape         : {img_orig.shape}")
print(f"Downscaled Shape       : {img_small.shape}")
print(f"Original Mean Intensity: {mean_orig:.4f}")
print(f"Restored Mean Intensity: {mean_restored:.4f}")
print(f"Absolute Mean Difference: {abs(mean_orig - mean_restored):.4f}")
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# ==============================================================================
# 1. SYNTHETIC IMAGE GENERATOR (To match screenshot)
# ==============================================================================
def create_sample_landscape(w, h):
    """Generates a landscape image (sky, sun, mountains) using NumPy and OpenCV."""
    img = np.zeros((h, w, 3), dtype=np.uint8)

    # Sky gradient (BGR)
    for y in range(h):
        r = int(90 + (140 - 90) * (y / h))
        g = int(120 + (160 - 120) * (y / h))
        b = int(160 + (200 - 160) * (y / h))
        img[y, :] = [b, g, r]

    # Yellow Sun
    sun_center = (int(w * 0.75), int(h * 0.22))
    sun_radius = int(min(w, h) * 0.08)
    cv.circle(img, sun_center, sun_radius, (70, 220, 250), -1, cv.LINE_AA)

    # Mountain Layers
    pts1 = np.array([[0, int(h*0.8)], [int(w*0.35), int(h*0.5)], [int(w*0.7), int(h*0.65)], [w, int(h*0.55)], [w, h], [0, h]], np.int32)
    cv.fillPoly(img, [pts1], (40, 90, 50))

    pts2 = np.array([[0, int(h*0.85)], [int(w*0.25), int(h*0.6)], [int(w*0.65), int(h*0.75)], [w, int(h*0.65)], [w, h], [0, h]], np.int32)
    cv.fillPoly(img, [pts2], (30, 70, 35))

    pts3 = np.array([[0, int(h*0.9)], [int(w*0.3), int(h*0.75)], [int(w*0.7), int(h*0.88)], [w, int(h*0.8)], [w, h], [0, h]], np.int32)
    cv.fillPoly(img, [pts3], (20, 50, 25))

    return img

# ==============================================================================
# 2. ADAPTIVE LABEL FUNCTION
# ==============================================================================
def add_adaptive_label(img_input, text="Dokumentasi PCVK – 2026"):
    img = img_input.copy()
    h, w = img.shape[:2]

    # Bar height formula
    bar_height = h // 12

    # Draw black bar
    cv.rectangle(img, (0, h - bar_height), (w, h), (0, 0, 0), -1)

    # Adaptive font parameters
    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = bar_height / 70.0
    thickness = max(1, int(bar_height / 35.0))

    # Adaptive text position
    text_x = int(w * 0.02)
    text_y = h - int(bar_height * 0.3)

    # Draw white text
    cv.putText(img, text, (text_x, text_y), font, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return img, bar_height

# ==============================================================================
# 3. DISPLAY ONE BY ONE
# ==============================================================================

# --- FIRST PAIR: 900 x 600 px (Landscape) ---
img1_bgr = create_sample_landscape(900, 600)
out1_bgr, bar1 = add_adaptive_label(img1_bgr)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(cv.cvtColor(img1_bgr, cv.COLOR_BGR2RGB))
axes[0].set_title("Masukan 900 x 600 px")
axes[0].axis('off')

axes[1].imshow(cv.cvtColor(out1_bgr, cv.COLOR_BGR2RGB))
axes[1].set_title(f"Keluaran: bar = 600//12 = {bar1} px")
axes[1].axis('off')

plt.tight_layout()
plt.show()

# --- SECOND PAIR: 480 x 760 px (Portrait) ---
img2_bgr = create_sample_landscape(480, 760)
out2_bgr, bar2 = add_adaptive_label(img2_bgr)

fig, axes = plt.subplots(1, 2, figsize=(8, 7))
axes[0].imshow(cv.cvtColor(img2_bgr, cv.COLOR_BGR2RGB))
axes[0].set_title("Masukan 480 x 760 px")
axes[0].axis('off')

axes[1].imshow(cv.cvtColor(out2_bgr, cv.COLOR_BGR2RGB))
axes[1].set_title(f"Keluaran: bar = 760//12 = {bar2} px")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

def create_square_thumbnail(img_input, bg_color=(255, 255, 255), store_text="SHOP"):
    """
    Converts any image into a 1:1 square marketplace thumbnail.
    - Preserves original aspect ratio.
    - Centers image on a solid background.
    - Dynamically adds a proportionally scaled store logo badge and border frame.
    """
    img = img_input.copy()
    h, w = img.shape[:2]

    # 1. Determine square canvas size (S = max dimension)
    S = max(h, w)

    # 2. Create solid color 1:1 square background canvas (Default: White BGR)
    canvas = np.full((S, S, 3), bg_color, dtype=np.uint8)

    # 3. Calculate offsets to center the image on canvas
    x_offset = (S - w) // 2
    y_offset = (S - h) // 2

    # 4. Place image onto the centered region of the canvas
    canvas[y_offset:y_offset + h, x_offset:x_offset + w] = img

    # 5. Add adaptive outer border frame (matching marketplace style)
    border_thick = max(1, int(S * 0.004))
    cv.rectangle(canvas, (0, 0), (S - 1, S - 1), (180, 180, 180), border_thick)

    # 6. Create Adaptive Store Logo Badge (Top-Right Corner)
    logo_radius = int(S * 0.045)
    logo_center_x = S - int(S * 0.07)
    logo_center_y = int(S * 0.07)

    # Badge outer ring & filled background
    cv.circle(canvas, (logo_center_x, logo_center_y), logo_radius, (30, 30, 30), -1, cv.LINE_AA)
    cv.circle(canvas, (logo_center_x, logo_center_y), logo_radius, (0, 215, 255), max(1, int(S * 0.003)), cv.LINE_AA) # Gold ring

    # Badge Text
    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = logo_radius / 35.0
    thickness = max(1, int(font_scale * 2))

    # Estimate text size for centering inside logo
    (text_w, text_h), _ = cv.getTextSize(store_text, font, font_scale, thickness)
    text_x = logo_center_x - (text_w // 2)
    text_y = logo_center_y + (text_h // 2)

    cv.putText(canvas, store_text, (text_x, text_y), font, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return canvas

# ==============================================================================
# DEMONSTRATION WITH LANDSCAPE AND PORTRAIT IMAGES
# ==============================================================================

# Helper function to generate synthetic sample images
def create_sample_image(w, h):
    img = np.zeros((h, w, 3), dtype=np.uint8)
    # Sky gradient
    for y in range(h):
        img[y, :] = [int(160 + 40*(y/h)), int(120 + 40*(y/h)), int(90 + 50*(y/h))]
    # Sun
    cv.circle(img, (int(w * 0.75), int(h * 0.25)), int(min(w, h) * 0.08), (70, 220, 250), -1, cv.LINE_AA)
    # Mountains
    pts = np.array([[0, int(h*0.8)], [int(w*0.35), int(h*0.5)], [w, int(h*0.65)], [w, h], [0, h]], np.int32)
    cv.fillPoly(img, [pts], (40, 90, 50))
    return img

# Create test images (Landscape & Portrait)
img_landscape = create_sample_image(900, 600)  # 3:2 Landscape
img_portrait  = create_sample_image(480, 760)  # Portrait

# Generate 1:1 Thumbnails
thumb_landscape = create_square_thumbnail(img_landscape, store_text="SHOP")
thumb_portrait  = create_square_thumbnail(img_portrait, store_text="SHOP")

# Display Results
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(cv.cvtColor(thumb_landscape, cv.COLOR_BGR2RGB))
axes[0].set_title(f"Landscape Output ({thumb_landscape.shape[1]}x{thumb_landscape.shape[0]} px)")
axes[0].axis('off')

axes[1].imshow(cv.cvtColor(thumb_portrait, cv.COLOR_BGR2RGB))
axes[1].set_title(f"Portrait Output ({thumb_portrait.shape[1]}x{thumb_portrait.shape[0]} px)")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
